```python
import re

def fix_email(email):
    if pd.isna(email):
        return email
    
    email = str(email).strip()

    # Agar @ already hai, email ko change mat karo
    if "@" in email:
        return email

    # Missing @ ko domain se pehle add karo
    domains = ["gmail.com", "yahoo.com", "outlook.com", "hotmail.com", "rediffmail.com"]

    for domain in domains:
        if email.endswith(domain):
            return email.replace(domain, "@" + domain)

    return email

customers["email"] = customers["email"].apply(fix_email)
```
